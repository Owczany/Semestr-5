```python
from math import comb, ceil

def ensemble(p, n):
    s = 0.0
    start = ceil((n+1)/2)
    succ_prob = p ** start
    loss_prob = (1 - p) ** (n - start)
    for i in range(start, n + 1):
        s += (comb(n, i) * succ_prob * loss_prob)
        succ_prob *= p
        loss_prob /= (1 - p)

    return s

print(ensemble(0.6, 8))
```
