```python
sentences = [
    ["to", "jest", "proste", "zdanie"],
    ["word2vec", "uczy", "się", "reprezentacji", "słów"],
    ["python", "jest", "popularny", "w", "uczeniu", "maszynowym"],
]

from gensim.models import Word2Vec

model = Word2Vec(
    sentences=sentences,
    vector_size=100,   # rozmiar wektora słowa
    window=5,          # zakres kontekstu
    min_count=1,       # minimalna liczba wystąpień słowa
    workers=4,         # liczba wątków
    sg=1               # 1 = skip-gram, 0 = CBOW
)
```

```python
words = ['python', 'uczy', 'popularny']
vectors = []

for word in words:
    vectors.append(model.wv[word])

print(vectors[0:3])
```
