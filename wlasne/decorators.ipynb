Czym ogólnie jest wrapper w pythonie

```python
# Simple decorator

def my_decorator(func):
    def wrapper():
        print('Początek funkcji')
        func()
        print('Koniec funkcji')
    return wrapper
```

```python
def say_hello():
    print('Cześć')

say_hello = my_decorator(say_hello)

say_hello()
```

```python
@my_decorator
def say_hi():
    print('Hi!')

say_hi()
```

```python
def decorator(func):
    def wrapper(*args, **kwargs):
        print('Coś robię przed funkcją')
        res = func(*args, **kwargs)
        print(res)
        print('Do something after function')
    return wrapper

@decorator
def add(a, b):
    return a + b

add(10, 10)
```

```python
def repeat(times):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for i in range(times):
                print(f'Zaczynam pętle nr {i + 1}')
                func(*args, **kwargs)
                print(f'Kończę pętle nr {i + 1}')
        return wrapper
    return decorator
    
@repeat(3)
def say(text):
    print(text)

say('Hello')

def what():
    print('what?')

what = repeat(5)(what)

what()
```

```python
def add(*args):
    return sum(args)

print(add(*[1, 2, 3]))
```

```python
def add_repr(cls):
    cls.__repr__ = lambda self: f"<{cls.__name__}>"
    return cls

@add_repr
class User:
    pass


u = User()

print(u)
```
