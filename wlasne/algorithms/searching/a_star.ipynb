```python
from collections import deque
import heapq
import numpy as np
import copy

board = [
    ['S',' ',' ',' ',' ',' ',' ',' ',' ',' '],
    [' ','X','X','X','X',' ',' ',' ',' ',' '],
    [' ','X',' ',' ',' ','W',' ',' ','M',' '],
    [' ','X',' ',' ',' ','W',' ',' ','M',' '],
    [' ','X',' ',' ',' ','W','X','X','X',' '],
    [' ','X','X','X','X',' ',' ',' ',' ',' '],
    [' ','W','W','W','X',' ','M','M',' ',' '],
    [' ',' ',' ',' ','X',' ','X','X','X',' '],
    [' ','X','X','X','X',' ',' ',' ',' ',' '],
    [' ','X',' ',' ',' ',' ',' ',' ',' ','G'],
]

HEIGHT, WIDTH = np.array(board).shape

MOVES = [(0, 1), (1, 0), (0, -1), (-1, 0)]

def manhattan_distance(start, end):
    xs, ys = start
    xe, ye = end

    return abs(xe - xs) + abs(ye - ys)

def a_star(starting_point = (0, 0), goal_point = (9, 9)):
    log = []
    pq = []
    checked = []
    
    heapq.heappush(pq, (0, starting_point))

    while pq:
        _, point = heapq.heappop(pq)
        x, y = point
        log.append((x, y))
        board[y][x] = 'V'
        checked.append(copy.deepcopy(board))

        if (x, y) == goal_point:
            return True, log, checked

        for dx, dy in MOVES:
            new_x = x + dx
            new_y = y + dy

            if 0 <= new_x < WIDTH and 0 <= new_y < HEIGHT:
                if board[new_y][new_x] != 'X' and board[new_y][new_x] != 'C' and board[new_y][new_x] != 'V':
                    heapq.heappush(pq, (manhattan_distance((new_x, new_y), goal_point), (new_x, new_y)))
                    board[new_y][new_x] = 'C'

    return False, log, checked

_, log, checked = a_star()

for i, p in enumerate(log):
    x, y = p
    board[y][x] = str(i)


# for row in board:
#     print(row)


for h in checked:
    for row in h:
        print(row)

    print()
```
