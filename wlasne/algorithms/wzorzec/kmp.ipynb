Algorithm KMP

```python
from typing import List

def longest_prefix_suffix(pattern: str) -> List[int]:
    m = len(pattern)
    i = 1
    j = 0
    lps = [0] * m

    while i < m:
        if pattern[i] == pattern[j]:
            j += 1
            lps[i] = j
            i += 1
        else:
            if j != 0:
                j = lps[j - 1]
            else:
                lps[i] = 0
                i += 1

    return lps

def kmp(text: str, pattern: str) -> List[int]:
    res = []

    lps = longest_prefix_suffix(pattern)

    i = 0
    j = 0

    while i < len(text):

        if pattern[j] == text[i]:
            i += 1
            j += 1
        else:
            if j != 0:
                j = lps[j - 1]
            else:
                i += 1

        if j == len(pattern):
            res.append(i - j)
            j = lps[j - 1]
            
    return res

print(kmp(text='aabaabaab', pattern='aab'))

print(longest_prefix_suffix('aabaabaaababba'))
```
