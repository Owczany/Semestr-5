```python
import numpy as np

# Przykładowa macierz 3x3
A = np.array([
    [2, 5, 8],
    [3, 4, 6],
    [7, 10, 1]
])

# Znajdujemy elementy parzyste i dzielimy je przez 2
A[A % 2 == 0] = A[A % 2 == 0] // 2  # Używamy // żeby wynik też był całkowity

print(A)
```
